# **GENERATOR**

Ditandai dengan yield, digunakan untuk menghemat RAM

```python
def hitungkuadrat(x):
    for i in range(1,x+1):
        yield i**2

for hasil in hitungkuadrat(5):
    print(hasil)
```

```python
def namabuah():
    buah = ['Pisang', 'Anggur', 'Kelapa', 'Nanas']
    for i in buah:
        yield i

for i in namabuah():
    print(i)
```
