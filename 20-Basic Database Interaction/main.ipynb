# **SIMPLE**

```python
import sqlite3

# Membuat koneksi ke database (jika belum akan ditambahkan)
conn = sqlite3.connect('Database1.db')

# Membuat Kursor
cursor = conn.cursor()

# Membuat tabel
cursor.execute('''CREATE TABLE mahasiswa (nama TEXT, umur INT, id INT)  ''')

# Menutup koneksi
conn.close()
```

#Contoh 2

```python
import sqlite3

def koneksi():  # Koneksikan dengan database yang ada
    return sqlite3.connect('Database1.db') 

def tambahdata(nama, umur):
    conn = koneksi()
    cursor = conn.cursor()
    cursor.execute("INSERT INTO mahasiswa (nama, umur) VALUES (?,?)", (nama, umur))
    conn.commit()
    conn.close()
    print(f"Mahasiswa {nama} berhasil ditambahkan")

if __name__ == "__main__":
    tambahdata('Asep', 60)
    tambahdata('Budi', 40)

def tambahlistdata(list_data):
    conn = koneksi()
    cursor = conn.cursor()
    cursor.executemany("INSERT INTO mahasiswa (nama, umur) VALUES (?,?)", (list_data))
    conn.commit()
    conn.close()
    print(f"Data {len(list_data)} berhasil dimasukkan")

if __name__ == "__main__":
    mahasiswa = [
        ('Iqbal', 19),
        ('Marsya', 20),
        ('Azka', 19),
        ('Fadil', 19)
    ]

tambahlistdata(mahasiswa)
```

# **INTERMEDIATE**

```python
import sqlite3

def koneksi():  # Koneksikan dengan database yang ada
    return sqlite3.connect('Database1.db') 

def lihatdata():
    conn = koneksi()
    cursor = conn.cursor()
    cursor.execute("SELECT * FROM mahasiswa")
    data = cursor.fetchall()
    conn.close()
    return data

for nama, umur in lihatdata():
    print(f"  - {nama} ({umur} tahun)")
```

**Menggabungkan Tabel**

Karena di tabel mahasiswa belum ada kolom id harus kita tambahkan dulu kolomnya

```python
import sqlite3

conn = sqlite3.connect('Database1.db')
cursor = conn.cursor()

def tambahrowid():
    cursor.execute("ALTER TABLE mahasiswa ADD COLUMN id INT")
    conn.commit()
    print("Kolom berhasil dibuat")

tambahrowid()
conn.close()
```

# Masukkan ID

```python
import sqlite3

conn = sqlite3.connect('Database1.db')
cursor = conn.cursor()

# Coba JOIN langsung seperti contoh Anda
cursor.execute("SELECT idmahasiswa.id, mahasiswa.nama FROM idmahasiswa JOIN mahasiswa ON idmahasiswa.id = mahasiswa.id")
data = cursor.fetchall()

if data:
    print("✅ JOIN BERHASIL! Data ditemukan:")
    for row in data:
        print(f"ID: {row[0]}, Nama: {row[1]}")
else:
    print("❌ JOIN TIDAK MENGHASILKAN DATA!")
    print("Kemungkinan: ID di tabel mahasiswa masih kosong (NULL)")
```

```python
import sqlite3

conn = sqlite3.connect('Database1.db')
cursor = conn.cursor()

# Ambil ID dari idmahasiswa (urutkan sesuai rowid di idmahasiswa)
cursor.execute("SELECT id FROM idmahasiswa")
ids = cursor.fetchall()

# Ambil data mahasiswa (urutkan sesuai rowid)
cursor.execute("SELECT rowid, nama FROM mahasiswa")
mhs = cursor.fetchall()

# Update ID di mahasiswa
print("=== Proses Update ID ===")
for i, (rowid, nama) in enumerate(mhs):
    if i < len(ids):
        id_baru = ids[i][0]
        cursor.execute("UPDATE mahasiswa SET id = ? WHERE rowid = ?", (id_baru, rowid))
        print(f"Update: {nama} → ID {id_baru}")

conn.commit()
print("\n✅ ID berhasil diisi!")

# Cek hasil update
cursor.execute("SELECT nama, umur, id FROM mahasiswa")
data = cursor.fetchall()
print("\n=== Data Mahasiswa Setelah Update ===")
for row in data:
    print(f"Nama: {row[0]}, Umur: {row[1]}, ID: {row[2]}")

conn.close()
```
