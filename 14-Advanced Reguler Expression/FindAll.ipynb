**FindAll()** mengembalikan semua kecocokan tanpa tumpang tindih dalam string sebagai daftar string

```python
import re

teks = "Ular melingkar mutar mutar di atas pagar"
teks_regex = r'([a-zA-Z]ar)'

cari = re.findall(teks_regex, teks)
print(cari)
```

```python
import re
courses = "Python:Programming; Java:Web; C++:System; "
items = re.findall("[^:]*:[^;]*;? ", courses)   # Selalu perhatikan spasi karena berpengaruh kecuali menggunakan (\s)
print(items)
```

# **ALTERNATION** 

```python
import re

teks = " Workshop akan dilaksanakan di Jakarta atau Bandung"
cari = re.search(r"dilaksanakan.*(Jakarta|Tangerang|Depok|Bekasi|Bandung)", teks)
print(cari.group())
```

# **SPLITING STRING**     


    Tanpa Menggunakan RegEx


```python
string = "Insya Allah akan terbuka sembilan belas juta lapangan pekerjaan"

string.split()
```

```python
string = "Insya;Allah;akan;terbuka;sembilan;belas;juta;lapangan;pekerjaan"

string.split(";", maxsplit=5)
```

```python
string = "Insya; \t;Allah;akan;terbuka;sembilan;belas;juta;lapangan;pekerjaan"
string.split(";",10)
```

```python
string = "Insya; \t;Allah;akan;terbuka;sembilan;belas;juta;lapangan;pekerjaan"
string.split(None,10)
```

**USING REGEX**

\W meaning WhiteSpaces

```python
import re
string = "Tidak ada yang lebih genting dibanding MBG bahkan lapangan kerja sekalipun"
re.split(r"\W+",string)
```

```python
import re

nama = "Surname: Iqbal, Prename: Hanafi, Professional: Blockchain Engineer"
print(re.split(r",* *\w*: ", nama))
```
