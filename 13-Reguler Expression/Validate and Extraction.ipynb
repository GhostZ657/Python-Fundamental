**VALIDATION** Using RegEx

```python
import re

email_pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'
email = input("Masukkan Email: ")

if re.match(email_pattern, email):
    print("{} Valid" .format(email))
else :
    print("{} Not Valid" .format(email))

```

**EXTRACTION** Using RegEx

```python
import re

email_pattern = r'@([\w\.-]+)'
email = "ghostballz657@gmail.com"
search = re.search(email_pattern,email)
print("Hasil Search Domain:", search.group())
print("Nama Domain:", search.group(1))
```

**VALIDATE** Using RegEx

```python
import re

number_phone = ["+62 812 9257 1633", "+1 415 555 2671", "20 7183 8750"]
number_pattern = r'^\+\d'

for number in number_phone:
    if re.search(number_pattern, number):
        print("This {} is Valid".format(number))
    else:
        print("This {} is Not Valid" .format(number))
```
