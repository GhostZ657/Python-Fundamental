**#1 MEMBUAT FILE**


```python
with open("sumber.txt","w") as file:
  file.write("Halo, ini adalah contoh file teks sederhana")

with open("sumber.txt","r") as file:
  baca = file.read()
  print(baca)
```

**#2 SALIN FILE**

```python
file_sumber = "sumber.txt"
file_tujuan = "tujuan.txt"

with open(file_sumber,"r") as sumber, open(file_tujuan,"w") as tujuan :
  for baris in sumber:
    tujuan.write(baris)
```

Alternatif Salin File

```python
import shutil
file_sumber = "sumber.txt"
file_tujuan = "tujuan.txt"
shutil.copyfile(file_sumber, file_tujuan)
```

**#3 READ, MANIPULATE, and DUMP**





```python
import csv

Data = [
    {'Nama':'Iqbal', 'Absen':'001', 'Nilai': 90},
    {'Nama':'Akhdan', 'Absen':'002', 'Nilai': 80},
    {'Nama':'Kela', 'Absen':'003', 'Nilai': 85},
    {'Nama':'Muti', 'Absen':'004', 'Nilai': 95},
]

# Definisikan variabel untuk file output
output_csv = "input.csv"

with open(output_csv, "w" ,newline="") as csvfile:
    Tabel = ['Nama','Absen','Nilai']
    writer = csv.DictWriter(csvfile, fieldnames=Tabel)
    
    # Buat Header
    writer.writeheader()
    
    # Buat perulangan untuk list Data
    for row in Data :
        writer.writerow(row)
```

```python
import csv

file_input = 'input.csv'
file_output = 'output.csv'

with open(file_input, 'r') as Masukan, open(file_output, 'w', newline="") as Keluaran:
    Pembaca = csv.reader(Masukan)   
    Penulis = csv.writer(Keluaran)

    # Hapus Header (Menyesuaikan dan memanipulasi data)
    next(Pembaca)
    
    # Lakukan pengulangan untuk setiap baris
    for baris in Pembaca:
        baruan_baris = [baris[0], int(baris[1]), baris[2]]    # Jadi baris ini mewakili value yang ada di list Data, walau ada dua value tetap menggunakan yang kita insialisasi seperti contoh 'baris'
        print(baruan_baris)
        Penulis.writerow(baruan_baris)

```

```python
import csv

file_input = 'input.csv'
file_output = 'output.csv'

with open(file_input, 'r') as Masukan, open(file_output, 'w', newline="") as Keluaran:
    Pembaca = csv.reader(Masukan)   
    Penulis = csv.writer(Keluaran)

    # Hapus Header (Menyesuaikan dan memanipulasi data)
    next(Pembaca)
    
    # Lakukan pengulangan untuk setiap baris
    for baris in Pembaca:
        if int(baris[2]) >= 85: # Selalu ubah dulu ke integer karena semua data kalau dari csv itu akan berbentuk string
            baruan_baris = [baris[0], int(baris[1]),int(baris[2])]    # Jadi baris ini mewakili value yang ada di list Data, walau ada dua value tetap menggunakan yang kita insialisasi seperti contoh 'baris'
            print(baruan_baris)
            Penulis.writerow(baruan_baris)

```
