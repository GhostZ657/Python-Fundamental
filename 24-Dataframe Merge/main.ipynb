# **CONCAT**

```python
import pandas as pd
import numpy as np

train_data = pd.read_csv("california_housing_train.csv")
test_data = pd.read_csv("california_housing_test.csv")
```

```python
train_data.shape
```

```python
test_data.shape
```

```python
pd1 = pd.DataFrame(train_data)
pd2 = pd.DataFrame(test_data)

print(pd1)
```

```python
print(pd2)
```

```python
total_kolom = pd.concat([pd1, pd2], axis=1)
print(total_kolom)
```

```python
total_baris = pd.concat([pd1, pd2], axis=0)
print(total_baris)
```

# **TABLE JOIN**

# *OUTER*

```python
DATA1 = {
    "Kota" : ['Jakarta', 'Bandung', 'Malang'],
    "Tahun" : [2019, 2020, 2021]
}

DATA2 = {
    "Kota" : ['Jakarta', 'Bandung', 'Las Vegas'],
    "Negara" : ['Idn', 'Idn', 'USA']
}

pd3 = pd.DataFrame(DATA1)
pd4 = pd.DataFrame(DATA2)

hasil_outer = pd.merge(pd3, pd4, on='Kota', how='outer')
print(hasil_outer)
```

# *INNER*

```python
hasil_inner = pd.merge(pd3, pd4, on='Kota', how='inner')
print(hasil_inner)
```

# *LEFT*

```python
hasil_left = pd.merge(pd3, pd4, on='Kota', how='left')
print(hasil_left)
```

# *RIGHT*

```python
hasil_right = pd.merge(pd3, pd4, on='Kota', how='right')
print(hasil_right)
```

# *MULTIPLE KEYS*

```python
import pandas as pd

# Contoh data dengan multiple keys
DATA1 = {
    "Kota" : ['Jakarta', 'Bandung', 'Malang', 'Jakarta'],
    "Tahun" : [2019, 2020, 2021, 2020],
    "Populasi" : [10.5, 2.5, 0.9, 10.6]
}

DATA2 = {
    "Kota" : ['Jakarta', 'Bandung', 'Jakarta', 'Surabaya'],
    "Tahun" : [2019, 2020, 2020, 2021],
    "Negara" : ['Idn', 'idn', 'Idn', 'Idn'],
    "Luas" : [661, 167, 661, 333]
}

df1 = pd.DataFrame(DATA1)
df2 = pd.DataFrame(DATA2)

print("DATA1:")
print(df1)
print("\nDATA2:")
print(df2)

# Join dengan multiple keys (Kota dan Tahun)
result_inner = pd.merge(df1, df2, on=['Kota', 'Tahun'], how='inner')
print("\nInner Join (Kota & Tahun):")
print(result_inner)
```
