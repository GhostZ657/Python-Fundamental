```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import(
    StandardScaler, MinMaxScaler, RobustScaler, LabelEncoder, OneHotEncoder, OrdinalEncoder
)
```

```python
data2 = {
    'produk':    ['A', 'B', 'C', 'D', 'E', 'F'],
    'kategori':  ['Elektronik', 'Fashion', 'Elektronik', 'Fashion', 'Elektronik', 'Fashion'],
    'rating':    ['Buruk', 'Cukup Buruk','Cukup', 'Cukup Bagus', 'Bagus', 'Sangat Bagus',],
    'harga':     [100, 5000, 800, 20000, 150, 35000],
    'terjual':   [1200, 30, 800, 15, 2000, 8]
}
df2 = pd.DataFrame(data2)
print(df2)
```

# **Tahap 0 Cek Skewness**

```python
print("Skewness Harga:",df2['harga'].skew())
print("Skewness Terjual:",df2['terjual'].skew())
```

# **Tahap 1 Transformation**

```python
# log_harga = data2['harga'].apply(np.log) # Tidak bisa digunakan
log_harga = np.log1p(df2['harga'])
print(log_harga)
log_terjual = np.log1p(df2['terjual'])
print(log_terjual)
```

```python
log_harga = pd.Series(log_harga) # Merubah ke bentuk pandas lagi
log_terjual = pd.Series(log_terjual) # Merubah ke bentuk pandas lagi
```

```python
print("Skewness harga SEBELUM log:", df2['harga'].skew().round(3))
print("Skewness harga SESUDAH log:", log_harga.skew().round(3))

print("Skewness terjual SEBELUM log:", df2['terjual'].skew().round(3))
print("Skewness terjual SETELAH log:", np.log1p(df2['terjual']).skew().round(3))

df2['log_harga'] = log_harga
df2['log_terjual'] = log_terjual
```

# **Tahap 2 Encoding**

*RATING ENCODE*

```python
df2['rating'] = df2['rating'].astype('category')
df2['rating_encode'] = df2['rating'].cat.codes
df2[['rating','rating_encode']]
```

```python
# Ga sesuai dengan yang saya inginkan 

rating_map = {
    'Buruk' : 0,
    'Cukup Buruk' : 1,
    'Cukup' : 2,
    'Cukup Bagus' : 3,
    'Bagus' : 4,
    'Sangat Bagus' : 5,
}

df2['rating_encode'] = df2['rating'].map(rating_map)
df2[['rating','rating_encode']]
```

*KATEGORI ENCODE*

```python
df2['kategori_enc'] = (df2['kategori']=='Elektronik').astype(int)

df2[['kategori','kategori_enc']]
```

```python
# Kolom produk bersifat unik tidak ada pola, jadi lebih baik untuk dihilangkan saja

# df2 = df2.drop(columns='produk')
print(df2.columns.tolist())
```

# **Tahap 3 Standardization**

Fokus kita disini ada pada kolom (log_harga, log_terjual)

```python
from sklearn.preprocessing import StandardScaler
std_scaler = StandardScaler()

df2[['log_harga_std','log_terjual_std']] = std_scaler.fit_transform(df2[['log_harga','log_terjual']])

print(df2[['log_harga', 'log_terjual', 'log_harga_std','log_terjual_std']].round(3))
```

```python
print("Mean:", df2[['log_harga_std', 'log_terjual_std']].mean().round(10).values)
print("Std :", df2[['log_harga_std', 'log_terjual_std']].std(ddof=0).round(10).values)
```

# **DATAFRAME FINAL**

```python
df_final = df2[['kategori_enc', 'rating_encode', 'log_harga_std', 'log_terjual_std']].copy()

pd.set_option('display.float_format', '{:.3f}'.format)
print(df_final)
print("\nShape:", df_final.shape)
print("Kolom:", df_final.columns.tolist())
```
