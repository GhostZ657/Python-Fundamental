**Matchin Pattern With Text**

**re.match()** : Fungsi ini mencoba mencocokkan pola dengan seluruh string. 

re.match(pattern, string, flag)

```python
import re

def FindMonthandYear(string):

    regex = r"([a-zA-Z]+) (\d+)"
    matching = re.match(regex,string)

    if matching ==  None:
        print("Hidup Jokowiiii!!!!")
        return # Hentikan fungsi di sini

    print("Month : %s" %(matching.group(0)))
    print("Year : %s" %(matching.group(1)))

FindMonthandYear("June 2026")
FindMonthandYear("Gaada yang lebih genting dari MBG")
```
