**Finding All Occurance In Pattern**

**re.findall()**: Mengembalikan semua kecocokan pola yang tidak tumpang tinding dalam string, sebagai daftar string

```python
import re

def NIMMahasigma(integer):
    regex = r"([a-zA-Z]+)(@)(\d+)"
    find = re.findall(regex,integer)
    match = re.match(regex,integer)
    print("NIM Mahasiswa:", find)
    print("Nama Mahasiswa:", match[1])
    print("Absen Mahasiswa:", match[3])

NIMMahasigma("iqbal@123")
```
