re.VERBOSE untuk menulis ekspresi reguler yang terlihat lebih bagus, mudah
dibaca, dan memungkinkan untuk memisahkan bagian logis dari pola secara visual
dan menambahkan komentar.

Spasi di dalam pola diabaikan, kecuali saat berada di kelas karakter, atau saat
didahului oleh garis miring terbalik yang tidak terhapus, atau di dalam token seperti
*? , ( ?: atau (?P.

Ketika sebuah baris berisi # yang tidak ada di kelas karakter dan tidak didahului
oleh garis miring terbalik, semua karakter dari # paling kiri hingga akhir baris
diabaikan.

```python
import re

def validationemail(email):
    regex_email = re.compile(r"""
    ^([a-z0-9_\.-]+) #Local Part
    @               # Single @ Sign
    ([a-z0-9_\.-]+) # Domain name
    \.              # Single dot
    ([a-z]{2,6})$   # Top level domain                
    """,re.VERBOSE | re.IGNORECASE)
    
    matching = regex_email.match(email)
    if matching == None :
            print("Email kamu {} tidak valid" .format(email))
    else:
            print("Email kamu {} valid" .format(email))

validationemail("ghostballz657@gmail.com")
validationemail("ghostballz657@gmail.com@")
```
