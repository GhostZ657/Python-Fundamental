# *PANDAS SERIES*

```python
import pandas as pd

s = [1, 3, 5]

myvar = pd.DataFrame(s)

print(myvar)
print("")
print(myvar[0])
```

```python
import pandas as pd

s = [1, 3, 5]

myvar = pd.Series(s, index = ['a', 'b', 'c'])

print(myvar)
print(myvar['a'])
```

# *Key/Value Object as Series*

```python
import pandas as pd

s = {'a':1, 'b':3, 'c':5}

myvar = pd.Series(s)

print(myvar)
print(myvar['a'])
```

*Memilih beberapa item dalam kamus*

```python
import pandas as pd

s = {'a':1, 'b':3, 'c':5}

myvar = pd.Series(s, index=['a', 'b'])

print(myvar)
print(myvar['a'])
```

# *Locate Row*

```python
import pandas as pd

data = {
    'Nama' : ['Asep', 'Budi', 'Caca'],
    'Umur' : [15, 14, 11]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.loc[0])
```

# *Named Index*

```python
import pandas as pd

data = {
    'Nama' : ['Asep', 'Budi', 'Caca'],
    'Umur' : [15, 14, 11]
}

df = pd.DataFrame(data, index=['Data 1', 'Data 2', 'Data 3'])

print(df)
```

*Located Named Index*

```python
print(df.loc['Data 1'])
```
