# **SIMPLE**

```python
import threading
import time

def import_number():
    for i in range(1,8):
        print(f'Number: {i}')
        time.sleep(1)

def import_text():
    for i in 'abcdefg':
        print(f"Letter: {i}")
        time.sleep(1.1)

thread1 = threading.Thread(target=import_number)
thread2 = threading.Thread(target=import_text)

thread1.start()
thread2.start()

thread1.join()
thread2.join()
```

#Contoh 2


```python
import threading
import queue

def antri(q):
    for i in range(5):
        q.put(i)
#       print(f"nomor:{i}")

my_queue = queue.Queue()
thread1 = threading.Thread(target=antri, args=(my_queue,))  # Gunakan koma karena parameternya membutuhkan Tuple 

thread1.start()

thread1.join()

print("\nIsi queue:")
while not my_queue.empty():
    print(my_queue.get())
```

# **INTERMEDIATE**

```python
import threading
import queue

kunci = threading.Lock()
def produsen(q):
    for i in range(int(input("Masukkan jumlah barang:"))):
        q.put(i)
        with kunci: 
            print(f'Barang {i}')

def konsumen(q):
    while True:
        item = q.get()
        if item is None:
            break
        with kunci:
            print("Dikonsumsi", item)

q = queue.Queue()
thread1 = threading.Thread(target=produsen, args=(q,))
thread2 = threading.Thread(target=konsumen, args=(q,))
thread1.start()
thread1.join()
print("Total Item: ",q.qsize()) # Untuk cek jumlah item 
thread2.start()
q.put(None) # Sinyal kepada konsumen untuk berhenti
thread2.join()
```

#Contoh 2

```python
from concurrent.futures import ThreadPoolExecutor
import time

def kuadrat(x):
    time.sleep(1)
    return x*x

data = [1, 2, 3, 4, 5]
with ThreadPoolExecutor(max_workers=3) as executor:
    hasil = list(executor.map(kuadrat, data))

print("hasil: ", hasil)
```

# **ADVANCE**

Eksekusi bersamaan dengan **ThreadPoolExecutor**

```python
from concurrent.futures import ThreadPoolExecutor, ProcessPoolExecutor
import time

def kuadrat(x):
    time.sleep(1)
    return x*x

data = [1, 2, 3, 4, 5]
with ThreadPoolExecutor(max_workers=3) as eksekutor :
    hasil = list(eksekutor.map(kuadrat, data)) 
print("Hasil :", hasil)
```

```python
from concurrent.futures import ProcessPoolExecutor

# Fungsi yang sangat sederhana
def kuadrat(x):
    return x * x

# Jalankan di dalam blok if __name__ == "__main__" (WAJIB di Windows/spawn)
if __name__ == "__main__":
    data = [1, 2, 3, 4, 5]
    with ProcessPoolExecutor(max_workers=3) as eksekutor:
        hasil = list(eksekutor.map(kuadrat, data))
    print("Hasil :", hasil)
```

```python
import multiprocessing as mp
from concurrent.futures import ProcessPoolExecutor
import traceback
import sys

def kuadrat(x):
    print(f"Processing: {x}")  # Debug
    return x * x

def worker_initializer():
    """Fungsi ini akan dijalankan di setiap worker"""
    print(f"Worker started with PID: {mp.current_process().pid}")

if __name__ == "__main__":
    print(f"Main PID: {mp.current_process().pid}")
    print(f"Start method: {mp.get_start_method()}")
    
    # Coba dengan 1 worker saja dulu
    try:
        with ProcessPoolExecutor(max_workers=1, initializer=worker_initializer) as executor:
            future = executor.submit(kuadrat, 5)
            result = future.result(timeout=5)  # Timeout 5 detik
            print(f"Result: {result}")
    except Exception as e:
        print(f"Error: {e}")
        traceback.print_exc()
```

```python
import multiprocessing as mp
from concurrent.futures import ProcessPoolExecutor
import sys
import traceback

def kuadrat(x):
    # Tambahkan logging untuk melihat apa yang terjadi
    print(f"Child process {mp.current_process().pid} menerima: {x}")
    return x * x

def test_sederhana():
    print(f"Main PID: {mp.current_process().pid}")
    print(f"Python version: {sys.version}")
    print(f"Start method: {mp.get_start_method()}")
    
    # Coba dengan 1 worker
    with ProcessPoolExecutor(max_workers=1) as executor:
        future = executor.submit(kuadrat, 5)
        try:
            result = future.result(timeout=10)
            print(f"Hasil: {result}")
        except Exception as e:
            print(f"Error: {e}")
            traceback.print_exc()

if __name__ == "__main__":
    test_sederhana()
```

```python
import multiprocessing as mp
from concurrent.futures import ProcessPoolExecutor

def test_func():
    return "Hello from child!"

if __name__ == "__main__":
    print(f"Main PID: {mp.current_process().pid}")
    
    with ProcessPoolExecutor(max_workers=1) as executor:
        future = executor.submit(test_func)
        result = future.result(timeout=5)
        print(f"Result: {result}")
```
