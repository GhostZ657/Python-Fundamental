# *CrossTab*

```python
percentage = pd.crosstab(index=flight_data['year'],
                        columns=flight_data['month'],
                        values=flight_data['passengers'],
                        aggfunc='max',
#                        margins=True,
#                        margins_name="Total",
#                        normalize='index'
)
```

```python
percentage
```

```python
percentage = pd.crosstab(index=flight_data['year'],
                        columns=flight_data['month'],
                        values=flight_data['passengers'],
                        aggfunc='sum').round(1)
percentage['Mean'] = percentage.mean(axis=1).round(0)
percentage['Median'] = percentage.median(axis=1).round(0)
percentage['Min'] = percentage.min(axis=1).round(0)
percentage['Max'] = percentage.max(axis=1).round(0)
percentage
```

```python
percentage.sum(axis=1)
```

# ***.pivot_table***

```python
flight_data.pivot_table( index = "year",
                        columns= "month",
                        values= "passengers",
#                        aggfunc=pd.Series.nunique,
                        aggfunc= "count"
)
```

# ***Melting***

```python
melting = percentage.reset_index().melt(id_vars=['year'],
                            var_name='Bulan',
                            value_vars=['Jan'],
                            value_name='Passangers')
melting
```
