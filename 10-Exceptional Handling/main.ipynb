**INTERMEDIATE: INVALID RESULT OF OPERATION**

#Contoh 1

```python
def devidenumber(a,b):
    try:
        result = a / b
        return("Hasil dari {}/{}={}".format(a,b,result))
    except ZeroDivisionError:
        return("Invalid result")
input_user = devidenumber(10,5)
print("Hasilnya:",input_user)
```

#Contoh 2

```python
def readfilecontent(filename):
    try:
        with open(filename,"r") as input:
            content = input.read()
            return(content)
    except FileNotFoundError:
        return("File is not existed!!!")
namafile = readfilecontent("not_excisted_file.txt")
print(namafile)
```

#Contoh 3

```python
try:
    num = int(input("Masukkan sebuah angka: "))
    result = 10 / num
    print("Hasil dari 10 dibagi {}".format(num),"Adalah: ", result)
except ZeroDivisionError:
    print("Masukkan angka selain 0")
except ValueError:
    print("Masukkan format angka yang benar")
```

**ADVANCE: HANDLE POSSIBLE ERROR FROM FILE**

```python
import json

def handleerror(filename):
    try:
        with open(filename, 'r') as input:
            output = json.load(input)
            print(output)
    except FileNotFoundError:
        return ("File Tidak Ditemukan!!!")
    except json.JSONDecodeError as e:
        return f"JSON decoding error {str(e)}"

filename = "Filejson.json"
output = handleerror(filename)
print("Output:", output)
```

#Buat Exception Sendiri (Custom Exception)

```python
class customexception(Exception):
    def __init__(self, massage):
        super().__init__(massage)

try:
    age = int(input("Masukkkan Umur:"))
    if age < 0 :
        raise customexception("Pls enter the invalid integer for age")
    print("You are {} years old".format(age))
except ZeroDivisionError:
    print("Masukkan angka selain 0")
except ValueError:
    print("Masukkan umur dengan format yang benar")
```
