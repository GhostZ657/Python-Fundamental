```python
class dog:
    def __init__(self,name,age):    #Bisa diberikan default argument
        self.name = name
        self.age = age

    def speak(self):
        print("Insya Allah akan {} buka sembilan belas juta lapangan padel".format(self.name))

    def speak(self):
        print("Insya Allah tahun ini akan berusia {} tahun".format(self.age))

Rodger = dog('Rodger',19)
Asep = dog('Asep',18)

print('My dog name is {}'.format(Rodger.name))
Rodger.speak()
Asep.speak()
```

# **INHERITANCE**

```python
class person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def display(self):
        print("Halo nama saya {}, umur saya {} tahun".format(self.name, self.age))

class mahasiswa:
    def __init__(self,name,age,nim,nilai):
        self.nim = nim
        self.nilai = nilai


        person.__init__(self , name, age)
    def details(self):
        print("Halo nama saya {}, umur saya {}, nim saya {}, nilai saya {}".format(self.name, self.age, self.nim, self.nilai))


iqbal = mahasiswa('iqbal',19,103062500009, 95)
iqbal.details()
```
