```python
class project:
    def __init__(self, nama, tanggal_mulai, tanggal_selesai):
        self.nama = nama
        self.tanggal_mulai = tanggal_mulai
        self.tanggal_selesai = tanggal_selesai

    def deadline(self):
        return ("Deadline {} tanggal {}".format(self.nama,self.tanggal_selesai))

class task:
    def __init__(self, nama, tanggal_mulai, tanggal_selesai, status="belum dikerjakan"):
        self.nama = nama
        self.start = tanggal_mulai
        self.end = tanggal_selesai
        self.status = status

    def marksascomplete(self):
        return ("{} telah selesai".format(self.nama))

from datetime import datetime
projek1 = project("Projek 1", datetime(2026, 10, 1), datetime(2026, 12, 1))
task1 = task("Projek 1", datetime(2026, 10, 1), datetime(2026, 12, 1))
print(projek1.deadline())
print(task1.marksascomplete())
print(task1.status)

```
