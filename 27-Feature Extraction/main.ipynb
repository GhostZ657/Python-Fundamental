# ***Math Operation***

```python
import pandas as pd

data_house = pd.read_csv("california_housing_train (1).csv")

data_house.head()
```

```python
other_rooms = data_house['total_rooms'] - data_house['total_bedrooms']

other_rooms
```

```python
data_house['other_rooms'] = other_rooms

data_house
```

```python
kepadatan = data_house['population'].div(data_house['households'])

kepadatan
```

```python
data_house = data_house.assign(kepadatan = kepadatan)
data_house
```

```python
total_ruang = data_house[['total_rooms','total_bedrooms','other_rooms']]
total_ruang
```
