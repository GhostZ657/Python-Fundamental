# **STANDARDIZATION**

```python
import seaborn as sns
import pandas as pd
import numpy as np
import sklearn
```

```python
dataset = sns.load_dataset('diamonds')

dataset.head()
```

```python
def min_max_scaled(x):
    result = (x-x.min()) / (x.max()-x.min())
    return result

dataset[['price','carat']].apply(min_max_scaled)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
min_max = MinMaxScaler()
hasil = min_max.fit_transform(dataset[['price','carat']])
print(hasil)
```

```python
pd.DataFrame(hasil, columns=[['price','carat']])
```

```python
def normal_scale(x):
    result = (x-x.mean() / x.std())
    return result
```

```python
dataset[['price','carat']].apply(normal_scale, axis=0)
```

```python
from sklearn.preprocessing import StandardScaler

norm_scaler = StandardScaler()
hasil2 = norm_scaler.fit_transform(dataset[['price', 'carat']])
```

```python
pd.DataFrame(hasil2, columns=[['Price', 'Carat']])
```
