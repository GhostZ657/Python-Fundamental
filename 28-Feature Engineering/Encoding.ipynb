```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
dataset = sns.load_dataset('diamonds')
```

```python
dataset.head()
```

```python
dataset['cut'] = dataset['cut'].astype('category')
dataset['cut_encodes'] = dataset['cut'].cat.codes
dataset[['cut','cut_encodes']]
```

```python
by_cut = dataset.groupby('cut')
Premium = by_cut.get_group('Premium')
banyak_premium = Premium.__len__()
print(f"Banyak diamonds dengan kualitas premium ada sebanyak {banyak_premium} unit")
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
cut_category = le.fit_transform(dataset['cut'])
```

```python
cut_category
```

```python
le.inverse_transform(cut_category)
```

```python
pd.get_dummies(dataset['cut'])
```
