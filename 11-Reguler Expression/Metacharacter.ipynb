**REGULER EXPRESSION**

```python
import re

A = "I'm student in mySkill"
search = re.search(r's',A)
print(search)
```

**METACHARACTER**

```python
import re

B = "Rifqi is student in UIN.Jakarta"   # Mencari karakter di sebelah kiri yang kita cari
search = re.search(r'.',B)
print(search)

search = re.search(r'\.',B)             # Mencari tepat karakter "."
print(search)
```

**SQUARE BRACKET([])**

```python
import re

B = "Rifqi is student in UIN.Jakarta"  
search = re.search(r'[UIN]',B)
print(search)
```

**CARET (^)**

```python
import re

B = "Rifqi is student in UIN.Jakarta"  
search = re.search(r'^Ri',B) #Hanya bisa memeriksa awalan string
print(search)

C = "Rifqi is student in UIN.Jakarta"  
search = re.search(r'^R',C) #Hanya bisa memeriksa awalan string
print(search)
```

**DOLLAR ($)**

```python
import re

B = "Rifqi is student in UIN.Jakarta"  
search = re.search(r'ta$',B) # Memeriksa AKhir String
print(search)

C = "Rifqi is student in UIN.Jakarta"  
search = re.search(r'a$',C) 
print(search)
```

**DOT (.)**

```python
import re

C = "Rifqi is student in UIN.Jakarta"  
search = re.search(r'i.i',C) 
print(search)

C = "Rifqi is student in UIN.Jakarta"  
search = re.search(r'U.N',C) 
print(search)
```

**OR (|)**

```python
import re

B = "Rifqi is student in UIN.Jakarta"  
search = re.search(r'R|i',B) # Memeriksa apakah di string ada R atau i
print(search)

C = "Rifqi is student in UIN.Jakarta"  
search = re.search(r'c|x',C) 
print(search)
```

**QUESTION MARK (?)**

```python
import re

B = "AABBCC"  
search = re.search(r'A?BC',B) 
print(search)

C = "ABBC"  
search = re.search(r'AB?C',C) 
print(search)

D = "ABCC"  
search = re.search(r'AB?C',D) 
print(search)

E = "AC"  
search = re.search(r'AB?C',E) 
print(search)

E = "ABDC"  
search = re.search(r'AB?C',E) 
print(search)

# Selama urutannya ABC akan selalu TRUE jika ada selipan variabel lain selain dari urutannya maka akan menjadi FALSE
# Misal ABDC itu akan menjadi FALSE/None
```

**STAR (*)**

cocok untuk nol atau lebih kemungkinan kemunculan variabel tersebut

```python
import re

B = "AACC"  
search = re.search(r'AB*C',B) 
print(search)

C = "ABBBBBBBC"  
search = re.search(r'AB*C',C) 
print(search)
```

**PLUS (+)**

Setidaknya kemunculan satu variabel atau lebih

```python
import re

B = "AACC"  
search = re.search(r'AB+C',B) 
print(search)

C = "ABBC"  
search = re.search(r'AB+C',C) 
print(search)
```

**BRACES ({A,B})**

Kondisi harus memenuhi kuantitas di dalamnya sebelah kiri minimal dan kanan maksimal variabel tersebut

```python
import re

B = "AACC"  
search = re.search(r'AB{1,3}C',B) 
print(search)

C = "ABBC"  
search = re.search(r'AB{1,4}C',C) 
print(search)

D = "ABBBBBC"  
search = re.search(r'AB{1,4}C',D) 
print(search)
```

**GROUP (a|b)**

Bisa mengoperasikan segala bentuk Regex untuk spesifik variabel di dalam kurung

```python
import re

B = "AACC"  
search = re.search(r'(A|B)C',B) 
print(search)

C = "ABBCD"  
search = re.search(r'(A|B)C',C) 
print(search)

D = "ABBBBBC"  
search = re.search(r'(A|B)CD',D) 
print(search)
```
