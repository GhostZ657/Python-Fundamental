**Re Function**

**re.findall()** Menampilkan semua kecocokan pola yang tidak tumpang tindih dalam string, sebagai daftar string

```python
import re

Sample = "Saya adalah mahasiswa dengan NIM 103062500009 dan teman saya dengan NIM 123456789"

# Harus menggunakan r sebelum value yang dicari
regex = r'\d+'
regex2 = r'\d'
find = re.findall(regex,Sample)
print(find)
find2 = re.findall(regex2,Sample)
print(find2)

# Alternatif
regex = '\\d+'
regex2 = '\\d'
find = re.findall(regex,Sample)
print(find)
find2 = re.findall(regex2,Sample)
print(find2)
```

**re.compile()** dimana reguler expression dikompilasi menjadi objek pola, yang memiliki metode untuk berbagai operasi seperti mencari kecocokan pola atau melakukan substitusi string

```python
import re

a = re.compile('[a-z]') # Selalu perhatikan lowercase/uppercase

print(a.findall("Saya suka makan tempe"))
```

**re.split()** Memisahkan string dengan pola tertentu

```python
from re import split

# Tanda baca termasuk karakter
print(split(r"\W","Saya suka membaca buku 'Rahasia Dunia yang Disembunyikan'"))

# Menghilangkan semua tanda baca
print(split(r"\W+","Saya suka membaca buku 'Rahasia Dunia yang Disembunyikan'"))
```

**re.sub(pola,pengganti,string,count,flag)** Mengganti value tertentu dengan suatu pengganti jumlah dan kondisi yang dikontrol

```python
from re import sub

print(sub('saya','****','Saya akan kembali ke kampung halaman saya',flags=re.IGNORECASE))
print(sub('saya','****','Saya akan kembali ke kampung halaman saya',count=1,flags=re.IGNORECASE))
print(sub('saya','****','Saya akan kembali ke kampung halaman saya'))
```

**re.subn()** Sama seperti sub() namun ada tambahan tuple total penggantian

```python
from re import subn

print(subn('saya','****','Saya akan kembali ke kampung halaman saya',flags=re.IGNORECASE))
print(subn('saya','****','Saya akan kembali ke kampung halaman saya',count=1,flags=re.IGNORECASE))
print(subn('saya','****','Saya akan kembali ke kampung halaman saya'))
```

**re.escape()** Mengamblikan substring degan tambahan backslash di depannya( \\ )

```python
from re import escape

print(escape('Saya akan kembali ke kampung halaman saya'))
print(escape('HI HI Hidupppp Jokowiiii'))
```

**re.search(pola,string)** Metode ini akan mengembalikan none jika tidak menemukan pola yang cocok, dan akan mengembalikan kecocokan yang pertama saja

```python
from re import search

regex = r"([a-zA-Z]+) (\d+)"

matching = search(regex, "I was born in May 06")

if matching != None:
    print("Match at index %s, %s" %(matching.start(), matching.end()))
    print("Full Matching : %s" %(matching.group(0)))
    print("First Matching : %s" %(matching.group(1)))
    print("Second Matching : %s" %(matching.group(2)))

if matching == None :
    print("Not a valid date")
```
