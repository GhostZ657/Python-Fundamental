**MATCH OBJECT**

```python
import re

s = "Saya akan lawan, hidup jokowi!!!"

res = re.search(r"\bJ",s)

if res:
    print(res.re)
else :
    print("Mas Aniez mas Aniez")

if res:
    print(res.string)
else :
    print("YNTKTS")
```
