# **GROUPING**

```python
import pandas as pd
import numpy as np
```

```python
house_data = pd.read_csv("House_Rent_Dataset (1).csv")
```

```python
house_data.head()
```

# *get_group*

Berfungsi untuk mengambil dataframe yang terdiri dari tiap kategori

```python
area_type = house_data.groupby('Area Type')
super_area = area_type.get_group('Super Area')
super_area.head(10)
```

# *TIPE AREA*

```python
house_data["Area Type"].unique()
```

```python
carpet_area = area_type.get_group('Carpet Area')
carpet_area.head()
```

# *by Area Locality*

```python
by_Area_Locality = house_data.groupby('Area Locality')
```

```python
South_Dum_Dum = by_Area_Locality.get_group('Keshtopur')
South_Dum_Dum.head()
```

# *MULTIPLE KEYS*

```python
bandel = by_Area_Locality.get_group('Bandel')
bandel.head()

by_area_and_local = house_data.groupby(['Area Type', 'Area Locality'])
by_area_and_local.get_group(('Super Area','Keshtopur'))
```

# **FUNCTION AGGREGATE**

```python
house_data.groupby('Area Type')[['Rent','Bathroom']].mean().sort_values(['Rent','Bathroom'],ascending=True)
```

```python
rent_by_Area = house_data.groupby('Area Type')['Rent'].agg(['min','max','median','mean'])
rent_by_Area
```

```python
house_data.groupby('City')['Rent'].agg(['min','max','median','mean'])
```

# *Multiple Keys Aggregate*

```python
house_data.groupby('City').agg({'Rent': ['min','max','median','mean'],'Size' : ['min','max']})
```

# **ITERATION**

```python
for group, df in by_area_and_local:
    print(f"{group[0] and group[1]} and number of rows {df.shape[0]}")
```
