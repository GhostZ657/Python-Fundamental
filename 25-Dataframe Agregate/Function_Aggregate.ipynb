```python
import pandas as pd
import numpy as np

# Membuat DataFrame contoh
data = {
    'Nama': ['Andi', 'Budi', 'Citra', 'Dewi', 'Eko'],
    'Jurusan': ['Informatika', 'Informatika', 'Sistem Informasi', 'Informatika', 'Sistem Informasi'],
    'Nilai_UAS': [85, 72, 90, 78, 88],
    'Nilai_Tugas': [80, 75, 85, 70, 92],
    'Kehadiran': [95, 88, 92, 78, np.nan]  # ada data kosong
}

df = pd.DataFrame(data)
print("DataFrame contoh:")
print(df)
```

```python
print("Rata-rata semua kolom numerik:")
print(df.mean(numeric_only=True))
print("\n" + "="*50)

print("Nilai terbesar setiap kolom: ")
print(df.max(numeric_only=True))
print("\n" + "="*50)

print("Nilai terkecil setiap kolom: ")
print(df.min(numeric_only=True))
print("\n" + "="*50)
```
